```python
# Draw as hypergraph using package 
import os
from   fnmatch import fnmatch
import numpy     as np
import hypernetx as hnx
import pandas    as pd
import networkx  as nx
import matplotlib.pyplot as plt
import helper_functions  as hf
```

```python
## Load files
OUTPUT_LOCATION = "../output/"
```

```python
print(f"Loading files from {OUTPUT_LOCATION}")

all_files = list() # tuples of ( filename_without_path, plotting_array, plt.figure )
for filename in os.listdir(OUTPUT_LOCATION):
    if fnmatch(filename, "*(h_graph)*.txt"):
        hypergraph = dict()
        hf.read_hypergraph_from_file(OUTPUT_LOCATION + filename, hypergraph)
        all_files.append( (filename, hypergraph, plt.subplots()) )
        plt.close()
print(f"Loaded {len(all_files)} file(s).")
```

```python
# Parse the data into a format for HNX to use
all_hnx_hypergraphs = list()
all_drawing_args    = list()
for (fname, orig_hypergraph, (fig, ax)) in all_files[10:]:
    hnx_hypergraph_d = dict()
    edge_labels      = dict()
    edge_colors      = []
    # Go through all hyperedges, with their repetitions
    n = 0
    plt.figure(fig)
    #ax  = fig.subplots() # default nrows=1, ncols=1
    ax.set_title(fname.rsplit('(')[0])
    for key in orig_hypergraph.keys():
        if len(key) < 2: 
            continue
        hnx_hypergraph_d[n] = list(key)
        count               = orig_hypergraph[key]
        edge_labels[n]      = f"{len(key)}:{count}" if count > 1 else None
        edge_colors.append(
            "grey"   if count < 2 else
            "blue"   if count < 3 else
            "cyan"   if count < 4 else
            "green"  if count < 5 else
            "orange" if count < 6 else
            "red"    
        )
        n += 1
    hnx_hypergraph = hnx.Hypergraph(hnx_hypergraph_d)
    nodes_kwargs   = {"facecolors":"navy"}
    edges_kwargs   = {"edgecolors":edge_colors} #, "alpha":(0.01)}#/(len(orig_hypergraph))+1)}
    edge_labels_kwargs = {"fontstyle":"oblique"}
    hnx.drawing.draw(hnx_hypergraph, with_node_labels=False, nodes_kwargs=nodes_kwargs, edge_labels=edge_labels, edges_kwargs=edges_kwargs, edge_labels_kwargs=edge_labels_kwargs)
    all_hnx_hypergraphs.append(hnx_hypergraph)
    all_drawing_args.append((edge_labels, edge_colors))
    
    fig.savefig(OUTPUT_LOCATION + fname.replace(".txt", ".svg").replace("history","h_figure"))
    #break # for testing purposes
print(f"Plotted {len(all_hnx_hypergraphs)} file(s).")
```
